```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from scipy.spatial.distance import pdist,squareform
from scipy.cluster.hierarchy import linkage,dendrogram,fcluster
```

```python
df = pd.read_csv("/content/fitness_health_tracking.csv")
print("first 5 records:")
print(df.head())
```

```python
print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nMissing value:")
print(df.isnull().sum())
```

```python
features =[
    'Daily_Steps',
    'Calories_Burned',
    'Workout_Duration',
    'Hours_Sleep'
]

x = df[features]

print('\nSelected Features:')
print(x.head())
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)

print('\nScaled Features:')
print(x_scaled[:5])
```

```python
x_scaled_Df = pd.DataFrame(x_scaled,columns=features)

print('\nScaled Features Dataframe:')
print(x_scaled_Df.head())
```

```python
x_scaled_Df =pd.DataFrame(
x_scaled,
columns=features
)

print("\nscaled Data:")
print(x_scaled_Df.head())
```

```python
distance = pdist(
    x_scaled_Df,
    metric='euclidean'
)

distance = squareform(distance)
print("\nDistance Matrix:")
print(distance)
```

```python
# Perform hierarchical clustering using Ward's linkage
Z = linkage(pdist(x_scaled_Df, metric='euclidean'), method='ward')

# Plot the Dendrogram
plt.figure(figsize=(12, 6))
dendrogram(
    Z,
    truncate_mode='lastp',
    p=30,
    show_leaf_counts=True,
    leaf_rotation=90.,
    leaf_font_size=12.,
    show_contracted=True
)
plt.title('Hierarchical Clustering Dendrogram (Ward Linkage)')
plt.xlabel('Cluster Size / Sample Index')
plt.ylabel('Distance')
plt.show()
```

```python
# Perform hierarchical clustering
Z = linkage(distance, method='ward')

# Plot the Dendrogram
plt.figure(figsize=(12, 6))
dendrogram(
    Z,
    truncate_mode='lastp',  # show only the last p merged clusters
    p=30,                   # show 30 leaf nodes
    show_leaf_counts=True,
    leaf_rotation=90.,
    leaf_font_size=12.,
    show_contracted=True
)
plt.title('Hierarchical Clustering Dendrogram (Ward Linkage)')
plt.xlabel('Cluster Size / Sample Index')
plt.ylabel('Distance')
plt.show()
```

```python
z = linkage(
    x_scaled_Df,
    method='single'
)

print("\nlinkage Matrix:")
print(z[:10])
```

```python
df['Cluster'] = fcluster(
    z,
    t=4,
    criterion ='maxclust'
)
```
